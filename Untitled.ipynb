```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import AgglomerativeClustering, DBSCAN, KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import calinski_harabasz_score, silhouette_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler 
```

```python
np.random.seed(42)
```

```python
X_raw, y_true = make_blobs(
    n_samples=500, centers=3, cluster_std=1.2, random_state=42
)
# Intentionally skew the feature scale to simulate raw real-world data
X_raw[:, 1] = X_raw[:, 1] * 50

print(f"Generated data shape: {X_raw.shape}")
print(f"Feature 1 Range: {X_raw[:, 0].min():.2f} to {X_raw[:, 0].max():.2f}")
print(f"Feature 2 Range: {X_raw[:, 1].min():.2f} to {X_raw[:, 1].max():.2f}")
```

```python
pipelines = {
    "K-Means": Pipeline(
        [
            ("scaler", StandardScaler()),
            ("clustering", KMeans(n_clusters=3, random_state=42)),
        ]
    ),
    "Hierarchical (Agglomerative)": Pipeline(
        [
            ("scaler", StandardScaler()),
            (
                "clustering",  # Fixed! Removed the accidental ",s"
                AgglomerativeClustering(n_clusters=3, linkage="ward"),
            ),
        ]
    ),
    "DBSCAN": Pipeline(
        [
            ("scaler", StandardScaler()),
            ("clustering", DBSCAN(eps=0.3, min_samples=7)),
        ]
    ),
}

# Dictionary to store cluster assignments
cluster_results = {}

# Fit the pipelines and retrieve labels
for name, pipeline in pipelines.items():
    pipeline.fit(X_raw)

    # Scikit-learn assigns results to the final step's 'labels_' attribute
    labels = pipeline.named_steps["clustering"].labels_
    cluster_results[name] = labels
```

```python
# ==========================================
# 5. VISUALIZATION (Fixed & Self-Contained)
# ==========================================

# 1. Ensure X_scaled is explicitly defined to prevent NameError
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

# 2. Initialize the side-by-side plot layout
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 3. Loop through and plot each algorithm's results
for ax, (name, labels) in zip(axes, cluster_results.items()):
    unique_labels = set(labels)
    colors = [
        plt.cm.Spectral(each) for each in np.linspace(0, 1, len(unique_labels))
    ]

    for k, col in zip(unique_labels, colors):
        if k == -1:
            col = [0, 0, 0, 1]  # Pure Black RGBA for DBSCAN noise/outliers

        class_member_mask = labels == k
        xy = X_scaled[class_member_mask]

        ax.scatter(
            xy[:, 0],
            xy[:, 1],
            c=[col],
            edgecolor="k",
            s=40,
            label=f"Cluster {k}" if k != -1 else "Noise",
        )

    ax.set_title(f"Clustering Results: {name}")
    ax.set_xlabel("Scaled Feature 1")
    ax.set_ylabel("Scaled Feature 2")
    ax.legend(loc="upper right", markerscale=0.6)

plt.tight_layout()
plt.show()
```
